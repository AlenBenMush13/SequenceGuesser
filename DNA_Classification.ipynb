## Interpretation note
The test accuracy is calculated from a random fragment-level split of two source genomes. Since fragments from the same genomes occur in both training and testing sets, this is not independent-genome validation. The model may largely learn nucleotide-composition / GC-content differences. The unknown sequence has no supplied true label, so its prediction cannot be independently verified.
